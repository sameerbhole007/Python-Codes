Assignment - if Statement

Q1. Take a number as input. If it is greater than 0, print "Positive Number".

```python
number=int(input("Enter a number"))
if number>=0:
    print("Positive Number!")
```

Q2. Take a person's age as input. If age is 18 or above, print "You are eligible to vote".

```python
age=int(input("Enter your age"))
if age>=18:
    print("You are eligible to vote!")
```

Q3. Take a number as input. If it is divisible by 3, print "Divisible by 3".

```python
num=int(input("Please enter a number"))
if num%3==0:
    print("Divisible by 3")
```

```python
marks=int(input("Enter student's marks"))
if marks>=40:
    print("Passed!")
```

Q5. Take a number as input. If it is even, print "Even Number". Print nothing if odd.

```python
number=int(input("Enter a number"))
if number%2==0:
    print("Even Number")
```

Q6. Take a string as input. If the length of the string is greater than 10, print "Long String".

```python
str=input("Enter a string")
if len(str)>10:
    print("Long String")
```

Q7. Take temperature as input. If temperature is above 37.5, print "You have fever. Please rest.".

```python
temperature=float(input("Enter Temperature"))
if temperature>37.5:
    print("You have fever. Please rest.")
```

Q8. Take a number as input. If it is a multiple of both 2 and 3, print "Multiple of 2 and 3".

```python
num=int(input("Enter a number"))
if num%2==0 and num%3==0:
    print("Multiple of 2 and 3")
```

Q9. Take a username as input. If username is "admin", print "Welcome Admin!".

Q10. Take a number as input. If the number is between 1 and 100 (inclusive), print "Valid Number".

```python
number=int(input("Enter a number:"))
if number>=1 and number<=100:
    print("Valid Number")
```
